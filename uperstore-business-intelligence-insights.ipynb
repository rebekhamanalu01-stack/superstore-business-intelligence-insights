```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams.update({
    'font.size': 11, 
    'axes.labelsize': 12, 
    'axes.titlesize': 14,
    'xtick.labelsize': 10, 
    'ytick.labelsize': 10, 
    'figure.figsize': (8, 4.5)
})
print("✓ Kotak 1 Sukses: Semua library data science siap digunakan!")
```

```python
nama_file = "UAS Statistika dan Probabilitas.xlsx"

try:
    df = pd.read_excel(nama_file)
    print("✓ Kotak 2 Sukses: Dataset berhasil dimuat!\n")
    print("3) 6 DATA PERTAMA (HEAD):")
    display(df.head(6))
    print("\n4) 6 DATA TERAKHIR (TAIL):")
    display(df.tail(6))
    print("\n5) STRUKTUR DATA (INFO):")
    df.info()
    print("\n6) RINGKASAN STATISTIK AWAL:")
    display(df.describe(include='all'))
    print(f"\n7) DIMENSI DATASET -> Jumlah Baris: {df.shape[0]}, Jumlah Kolom: {df.shape[1]}")
except Exception as e:
    print(f"❌ KENDALA: File tidak ditemukan. Pastikan file Excel ditaruh satu folder dengan file notebook ini. Detail: {e}")
```

```python
print("=== TAHAP 2: STATISTIK DESKRIPTIF ===")
rata_sales = df['Sales'].mean()
print(f"1) Rata-rata Penjualan: {rata_sales:.2f}\n")
print("2) Statistik Deskriptif Kolom Sales:")
print(f"   - Minimum Penjualan  : {df['Sales'].min():.2f}")
print(f"   - Maksimum Penjualan : {df['Sales'].max():.2f}")
print(f"   - Median Penjualan   : {df['Sales'].median():.2f}")
print(f"   - Standar Deviasi    : {df['Sales'].std():.2f}")
print(f"   - Kuartil 1 (Q1)     : {df['Sales'].quantile(0.25):.2f}")
print(f"   - Kuartil 3 (Q3)     : {df['Sales'].quantile(0.75):.2f}\n")

print("3) Top 10 Order ID dengan Penjualan Terbesar:")
display(df.groupby('Order ID')['Sales'].sum().nlargest(10).reset_index())

kategori_kolom = ['Ship Mode', 'Segment', 'Region', 'Category', 'Sub-Category', 'Order Priority']
print("\n5) Frekuensi Transaksi & 6) Total Sales per Kategori:")
for col in kategori_kolom:
    print(f"\n--- Analisis Kategori: {col} ---")
    freq = df[col].value_counts().to_frame(name='Jumlah Transaksi')
    sales = df.groupby(col)['Sales'].sum().sort_values(ascending=False).to_frame(name='Total Sales ($)')
    display(pd.concat([freq, sales], axis=1))
```

```python
print("=== TAHAP 4: ANALISIS PENJUALAN BERDASARKAN PRODUK ===")
cat_sales = df.groupby('Category')['Sales'].sum().reset_index()
sns.barplot(data=cat_sales, x='Category', y='Sales', palette='deep', hue='Category', legend=False)
plt.title('Total Penjualan Berdasarkan Kategori Utama Produk')
plt.show()

print("\n9) Total Profit per Category:")
display(df.groupby('Category')['Profit'].sum().to_frame())
print("\n10) Total Profit per Sub-Category:")
display(df.groupby('Sub-Category')['Profit'].sum().sort_values(ascending=False).to_frame())
```

```python
print("=== TAHAP 5: VISUALISASI VARIABEL KATEGORI GANDA ===")
sns.barplot(data=df, x='Region', y='Sales', hue='Category', estimator=sum, errorbar=None, palette='bright')
plt.title('Kombinasi Wilayah Makro (Region) dan Kategori Terhadap Sales')
plt.xticks(rotation=30)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()

sns.barplot(data=df.groupby('Order Priority')['Profit'].sum().reset_index(), x='Order Priority', y='Profit', palette='cool', hue='Order Priority', legend=False)
plt.title('Evaluasi Tingkat Urgensi Transaksi Terhadap Total Profit')
plt.show()
```
