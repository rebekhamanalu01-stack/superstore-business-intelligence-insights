```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)
```

```python
nama_file = "UAS Statistika dan Probabilitas.xlsx"

try:
    df = pd.read_excel(nama_file)
    print("✓ Kotak 2 Sukses: Dataset berhasil dimuat!\n")
    print("3) 6 DATA PERTAMA (HEAD):")
    display(df.head(6))
    print("\n4) 6 DATA TERAKHIR (TAIL):")
    display(df.tail(6))
    print("\n5) STRUKTUR DATA (INFO):")
    df.info()
    print("\n6) RINGKASAN STATISTIK AWAL:")
    display(df.describe(include='all'))
    print(f"\n7) DIMENSI DATASET -> Jumlah Baris: {df.shape[0]}, Jumlah Kolom: {df.shape[1]}")
except Exception as e:
    print(f"❌ KENDALA: File tidak ditemukan. Pastikan file Excel ditaruh satu folder dengan file notebook ini. Detail: {e}")
```

```python
print("=== TAHAP 2: STATISTIK DESKRIPTIF ===")
rata_sales = df['Sales'].mean()
print(f"1) Rata-rata Penjualan: {rata_sales:.2f}\n")
print("2) Statistik Deskriptif Kolom Sales:")
print(f"   - Minimum Penjualan  : {df['Sales'].min():.2f}")
print(f"   - Maksimum Penjualan : {df['Sales'].max():.2f}")
print(f"   - Median Penjualan   : {df['Sales'].median():.2f}")
print(f"   - Standar Deviasi    : {df['Sales'].std():.2f}")
print(f"   - Kuartil 1 (Q1)     : {df['Sales'].quantile(0.25):.2f}")
print(f"   - Kuartil 3 (Q3)     : {df['Sales'].quantile(0.75):.2f}\n")

print("3) Top 10 Order ID dengan Penjualan Terbesar:")
display(df.groupby('Order ID')['Sales'].sum().nlargest(10).reset_index())
print("4) Top 10 Order ID dengan Penjualan Terkecil:")
display(df.groupby('Order ID')['Sales'].sum().nsmallest(10).reset_index())

kategori_kolom = ['Ship Mode', 'Segment', 'Region', 'Category', 'Sub-Category', 'Order Priority']
print("\n5) Frekuensi Transaksi & 6) Total Sales per Kategori:")
for col in kategori_kolom:
    print(f"\n--- Analisis Kategori: {col} ---")
    freq = df[col].value_counts().to_frame(name='Jumlah Transaksi')
    sales = df.groupby(col)['Sales'].sum().sort_values(ascending=False).to_frame(name='Total Sales ($)')
    display(pd.concat([freq, sales], axis=1))
```

```python
print("6a) Total Penjualan Berdasarkan Ship Mode:")
display(df.groupby('Ship Mode')['Sales'].sum().reset_index())

print("\n6b) Total Penjualan Berdasarkan Segment:")
display(df.groupby('Segment')['Sales'].sum().reset_index())

print("\n6c) Total Penjualan Berdasarkan Region:")
display(df.groupby('Region')['Sales'].sum().reset_index())

print("\n6d) Total Penjualan Berdasarkan Category:")
display(df.groupby('Category')['Sales'].sum().reset_index())

print("\n6e) Total Penjualan Berdasarkan Sub-Category:")
display(df.groupby('Sub-Category')['Sales'].sum().reset_index())

print("\n6f) Total Penjualan Berdasarkan Order Priority:")
display(df.groupby('Order Priority')['Sales'].sum().reset_index())
```

```python

print("7a) Statistik Quantity:")
display(df['Quantity'].agg(['sum', 'mean', 'min', 'max']).to_frame())

print("\n7b) Statistik Discount:")
display(df['Discount'].agg(['sum', 'mean', 'min', 'max']).to_frame())

print("\n7c) Statistik Profit:")
display(df['Profit'].agg(['sum', 'mean', 'min', 'max']).to_frame())

print("\n7d) Statistik Shipping Cost:")
display(df['Shipping Cost'].agg(['sum', 'mean', 'min', 'max']).to_frame())
```
